```python
import torch

embedding = torch.nn.Embedding(
    num_embeddings=5,
    embedding_dim=3
)

print("Embedding Matrix:")
print(embedding.weight)

tokens = torch.tensor([0, 2, 4])

vectors = embedding(tokens)

print("\nToken IDs:")
print(tokens)

print("\nEmbedding Vectors:")
print(vectors)

print("\nOutput Shape:")
print(vectors.shape)
```
